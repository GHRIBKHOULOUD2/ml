```python
import matplotlib.pyplot as plt
X = [-1, 0, 1, 2]  
Y = [3, 2, 4, 1]   
plt.plot(X, Y,'b.')          
plt.grid()
plt.axis([-2, 3, 0, 5])
plt.xlabel("mes abscisses")   
plt.ylabel("mes ordonnées")  
plt.title("courbe d’équation $y =f(x)$") 
plt.show()
```

```python
import pandas as pd
import numpy as np


years = np.random.randint(1, 11, size=40)
salary = 2000 + 200 * years + np.random.randint(1, 401, size=40)
df = pd.DataFrame(
    {
        "years":years,
    "salary":salary
    }
)
print(df.head())
df.to_csv("kh",index=False)
print(2000 + 200 * 10 + 200
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("kh")
plt.scatter(data["years"],data["salary"])
plt.xlabel("years")
plt.ylabel("salary")
plt.title("salary vs years of experience")
plt.show()
```
